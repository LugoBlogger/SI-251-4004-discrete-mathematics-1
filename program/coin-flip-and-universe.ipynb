Estimate the number of atoms in the observable universe.

1. Count the Galaxies
   - Telescope map the sky.
   - They count visible galaxies.
   - The estimate is **2 trillion galaxies** in our observable universe.

2. Find the Avereage Mass of a Galaxy
   - Scientists weight galaxies.
   - They track how fast galaxies rotate.
   - The average galaxy has about **100 billion stars**.
   - This equals roughly $2 \times 10^{41}$ kilograms of matter per galaxy.

3. Calculate Total Ordinary Mass
   - Multiply the galaxies by the galaxy mass.
   - Total cosmic matter equals roughly **$10^{53}$ kilograms**.
   - This only includes normal matter (starts, gas, dust).
   - It excludes dark matter.

4. Divide by the Mass of Hydrogen
   - Over 90% of atoms are hydrogen.
   - One hydrogen atom weighs $1.67 \times 10^{-27}$ kilograms.
   - Divide total mass ($10^{53}$) by hydrogen mass ($1.67 \times 10^{-27}$ kg)
   - The final result is roughly **$10^{80}$ atoms**
